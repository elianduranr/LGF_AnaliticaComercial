## Interpretacion para el modelo

El pronostico no debe modelarse solo como una serie total. La demanda cambia por color y producto-color; algunos colores se mueven juntos semanalmente, mientras otros tienen correlaciones bajas o negativas, lo que sugiere patrones de sustitucion, temporadas o mercados distintos. Por eso la siguiente version del modelo deberia usar semana, color, producto, pais y cliente como variables de segmentacion, y comparar baseline reciente contra estacionalidad anual.
